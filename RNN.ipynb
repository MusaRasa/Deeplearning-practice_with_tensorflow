```python
import tensorflow as tf
```

# This is the RNN formula
RNN = Whh*Ht-1+Wxh*Xt+b

1. Xt = input
2. Ht-1 = previous hidden state
3. Whh and Wxh = Weight matrices
4. b = Bias vector


```python
Xt = tf.constant([1.0,0])
Ht_1 = tf.constant([0.5,-0.5])
Whh = tf.constant([[0.1,0.2],[-0.1,0.3]])
Wxh = tf.constant([[0.4,0.5],[0.6,0.7]])
Bias = tf.constant([0.1,0.2])
mult1 = tf.multiply(Whh,Ht_1)
multi2 = tf.multiply(Wxh,Xt)

Rnn = mult1 +multi2 +Bias
print(Rnn)
```
