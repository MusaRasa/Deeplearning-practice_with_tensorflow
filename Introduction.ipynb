### Different operation

```python
a = tf.constant([1,2])
b = tf.constant([3,4])
a,b
```

```python
# Sum operation
ResSum = tf.add(a,b)
ResSum
```

```python
# Subtract peration
Sub = tf.subtract(a,b)
Sub
```

```python
# The multiplication 
ev = tf.multiply(a,b)
ev
```

```python
a_1 = tf.constant([[1,2],[4,3]])
b_1 = tf.constant([[4,3],[8,7]])
mm = tf.matmul(a_1,b_1)
mm
```

```python
# Division 
a,b
```

```python
division = tf.divide(a_1,b_1) # To tensorflow the number/0 is infinit
division
```
