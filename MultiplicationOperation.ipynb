# Min, Max, Median, STD
 

```python
am = tf.constant([[1,2,4,5],[9,0,8,7]],dtype = float)
min = tf.reduce_min(am)
min
```

```python
# Max 
max = tf.reduce_max(am,axis= None)
print(am)
print(max)
```

```python
# Mean 
mean = tf.reduce_mean(am)
mean.numpy()
```

```python
sum = tf.reduce_sum(am)
sum
```

```python
# Standard division 
std = tf.math.reduce_std(am)
print(std)
print(f"STD: {std:.2f}")
```

```python
# Sum 
sum = tf.reduce_sum(am)
print("SUM: ",sum)
```
