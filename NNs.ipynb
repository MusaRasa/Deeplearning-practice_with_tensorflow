# Nural network with tensorflow and keras

```python
import tensorflow as tf
from tensorflow import keras
```

### Data definition
+ At the first we implemnt our data fo linear regression formula
+ Linear regression Formula: y = ax+b
+ a is a bais
+ b is variance to our model
+ x is a vertor

```python
# initialization of x, y
x = tf.constant([-1,-2,0,1,2,3,4,5,6,7,5,4,2,4,3,4,4,5,5,6,1,2,3,4,5,3,4,5,6,6,7,4,3,2,3,4,4,5,5,6,6,7,9],dtype = float)
y = 3*x  -5
```

```python
x.numpy()
```

```python
y.numpy()
```

# Model definition

```python
model = keras.Sequential(
    [
        keras.layers.Dense(units=1,activation="relu",input_shape = [1])
        
    ]
)
```

```python
model.compile(optimizer = 'sgd',loss = 'mean_squared_error')
model.summary()
```

```python
model.fit(x,y,batch_size=1,epochs = 120)
```

Test our model

```python
x_test = tf.constant([10,11,17,12,70,29],dtype = float)
y_test = x_test*3 - 5
y_test.numpy()
```

```python
# The testing prediction our model
y_pred = model.predict(x_test)
y_pred
```

```python
import matplotlib.pyplot as plt
plt.scatter(x_test.numpy(), y_test.numpy(),color = 'red',linewidths = 8)
plt.scatter(x_test.numpy(), y_pred,color = 'blue',linewidths = 2)
```
